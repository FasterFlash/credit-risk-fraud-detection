```python
CATALOG_PATH = "credit_risk_fraud_detection"
SCHEMA_PATH = "bronze"
PATH = "f/{CATALOG_PATH}.{SCHEMA_PATH}"
```

```python
BUREAU_T = "bureau_pulls"

query = f"SELECT * FROM {CATALOG_PATH}.{SCHEMA_PATH}.{BUREAU_T}"
spark.sql(query).display()
```

```python
query = f"SELECT * FROM {CATALOG_PATH}.{SCHEMA_PATH}.{BUREAU_T} WHERE _rescued_data is not null"
spark.sql(query).display()
```

```python
query = f"SELECT bureau_name, COUNT(*) as t_count FROM {CATALOG_PATH}.{SCHEMA_PATH}.{BUREAU_T} group by bureau_name order by t_count desc"
spark.sql(query).display()
```
