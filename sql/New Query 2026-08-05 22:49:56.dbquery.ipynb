```sql
CREATE OR REPLACE FUNCTION credit_risk_fraud_detection.ml.get_dq_summary()
RETURNS TABLE (violation_type STRING, violation_count BIGINT, pct_of_all_violations DOUBLE)
COMMENT 'Returns a summary of current data quality violations detected in the pipeline, broken down by type. Use when asked about data quality issues, DLQ, or DQ violations.'
RETURN
  SELECT violation_type, violation_count, pct_of_all_violations
  FROM credit_risk_fraud_detection.gold.kpi_dq_violation_summary
  ORDER BY violation_count DESC;
```
