```sql
CREATE OR REPLACE FUNCTION credit_risk_fraud_detection.ml.get_loan_risk_score(
    loan_id STRING COMMENT 'The loan_account_id to look up the current delinquency escalation risk score for, e.g. LN_XXXXXXXX'
)
RETURNS TABLE (loan_account_id STRING, escalation_probability DOUBLE, loan_status_at_event STRING, dpd_at_event INT)
COMMENT 'Returns the most recent delinquency escalation risk score for a given loan_account_id.'
RETURN
  SELECT loan_account_id, escalation_probability, loan_status_at_event, dpd_at_event
  FROM credit_risk_fraud_detection.gold.delinquency_risk_scores
  WHERE loan_account_id = loan_id;
```
