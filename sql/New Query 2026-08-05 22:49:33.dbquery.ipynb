```sql
CREATE OR REPLACE FUNCTION credit_risk_fraud_detection.ml.get_fraud_score(
    txn_id STRING COMMENT 'The transaction_id to look up the fraud probability score for, e.g. TXN_XXXXXXXX'
)
RETURNS TABLE (transaction_id STRING, amount DOUBLE, fraud_probability DOUBLE, fraud_flag_predicted BOOLEAN, is_fraud BOOLEAN)
COMMENT 'Returns the predicted fraud probability and actual fraud label for a given transaction_id.'
RETURN
  SELECT transaction_id, amount, fraud_probability, fraud_flag_predicted, is_fraud
  FROM credit_risk_fraud_detection.gold.fraud_predictions
  WHERE transaction_id = txn_id;
```
