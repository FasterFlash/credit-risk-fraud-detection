```sql
CREATE OR REPLACE FUNCTION credit_risk_fraud_detection.ml.get_portfolio_summary()
RETURNS TABLE (metric STRING, value STRING)
COMMENT 'Returns high-level portfolio KPIs: overall default rate, fraud incidence rate, and NPA rate by persona. Use for general portfolio-health questions, not single-loan or single-transaction questions.'
RETURN
  SELECT 'overall_default_rate' as metric, cast(default_rate as string) as value
  FROM credit_risk_fraud_detection.gold.kpi_portfolio_default_rate WHERE loan_type = 'ALL'
  UNION ALL
  SELECT 'overall_fraud_incidence_rate', cast(fraud_incidence_rate as string)
  FROM credit_risk_fraud_detection.gold.kpi_fraud_incidence WHERE fraud_type = 'ALL'
  UNION ALL
  SELECT concat('npa_rate_', persona), cast(npa_rate as string)
  FROM credit_risk_fraud_detection.gold.kpi_npa_by_persona;
```
